# Context-first lattice analysis

Read the original texts before consulting category labels. This notebook
replays saved evidence and numerical audits, not fresh model inference.
The qualitative choices are exploratory and not analyst-blinded.
See `experiments/activation_studies/contextreading_v1/READINGS.md`.

A join intersects extents; a meet weakens requirements and can admit extra
items. FG closes a numerical description while preserving its extent.
None of these laws by itself assigns semantic meaning.


```python
from pathlib import Path
import json
import os
import sys

root = Path(os.environ.get("LATTICE_REPOSITORY", Path.cwd())).resolve()
while not (root / "src/lattmc").exists() and root != root.parent:
    root = root.parent
if not (root / "src/lattmc").exists():
    raise RuntimeError("Set LATTICE_REPOSITORY to the checkout root.")
sys.path[:0] = [str(root), str(root / "src")]
out = Path(os.environ.get(
    "LATTICE_READING_RESULTS",
    root / "data/activation_studies/contextreading_v1"))
print("Interpreter:", sys.executable)
print("Records:", out)
```

```python
from lattmc.contextstudy.semantic_audit_codexgen import audit

print(audit(root, out))
```

## The complete operation grid

Every full and selected query is retained, including zero meets and empty
joins. Counts use 25,600 original GPT-2 chunks, including legacy BOS.
Four source rows are refreshed in memory; other cached rows are unchanged.
These chunks are not independent original documents.


```python
import pandas as pd
from IPython.display import HTML, display

rows = []
for kind in ("sae", "tc"):
    for block in (0, 8, 11):
        data = json.loads((out / f"gpt2_{kind}{block}.json").read_text())
        for r in data["records"]:
            rows.append({
                "condition": f"{kind}{block}", "group": r["group"],
                "rule": r["rule"],
                "constituents": [c["count"] for c in r["components"]],
                "meet": r["meet"]["count"], "join": r["join"]["count"],
                "additional meet": r["extra_count"],
                "closed meet coordinates": r["meet"]["closed_positive"],
            })
frame = pd.DataFrame(rows)
print("Complete grid shape:", frame.shape)
display(frame[frame["condition"] == "tc11"])
```

## Original texts and exact witnesses

Rose marks any positive-coordinate witness. Bold rose marks a token that
satisfies the entire query. Yellow D marks collective item satisfaction
without any whole-query token. These are numerical, not semantic labels.
The full original sequence decoding is printed below each token display;
separately decoded UTF-8 byte fragments may otherwise show replacements.
Queries retain recorded amplitudes, with no membership epsilon.


```python
from lattmc.contextstudy.witnessrender_codexgen import snippet

choices = [
    ("tc11", "research_three", "full", "meet", 16621),
    ("tc11", "mixed_pair", "rank1", "join", 9155),
    ("tc11", "biological_four", "rank1", "join", 23609),
    ("tc11", "mixed_three", "rank3", "join", 12902),
    ("tc11", "cat_dog", "full", "meet", 17545),
    ("sae11", "new_de", "rank23", "join", 21747),
]
for condition, group, rule, op, row in choices:
    data = json.loads((out / f"{condition}_witnesses.json").read_text())
    r = dict(next(r for r in data["records"] if r["row"] == row
                  and r["group"] == group and r["rule"] == rule
                  and r["operation"] == op))
    r["valid_positions"] = list(range(128))
    r["diagnostic"] = (r["partial"] or [0])[0]
    badge = r["status"]
    if badge == "D":
        badge = '<mark style="background:#FFEF9F">D</mark>'
    display(HTML(f'<h4>{condition}: {group}, {rule} {op}, row {row}</h4>'
                 + badge + '<p>' + snippet(r, as_html=True) + '</p>'))
    print("Complete original decode:", r["raw_text"])
    print("Extent:", r["extent"], "Exact query:", r["query"])
    print("Coordinate inequalities:", r["inequalities"])
```

## Structural readings before the later label comparison

Pythia's early-period pattern includes abbreviation periods, not just title
separators. SmolLM2's name-fragment pattern has a year-digit counterexample.
Members and nonmembers were sampled without new category selection. Earlier
source labels were known; this is not a blinded semantic precision study.


```python
data = json.loads((out / "structure_probe.json").read_text())
for family, row in [("pythia_topk", 1723), ("smol_topk", 1810)]:
    r = dict(next(r for r in data["records"] if
                  r["family"] == family and r["row"] == row))
    r.update(valid_positions=r["valid"], partial=r["whole"], status="S")
    r["diagnostic"] = r["whole"][0]
    display(HTML(f'<h4>{family}, row {row}</h4><p>'
                 + snippet(r, as_html=True) + '</p>'))
    print(r["raw_text"])
    print("Whole-query positions:", r["whole"], "Query:", r["query"])
notes = json.loads((out / "interpretations_before_labels.json").read_text())
print("Recorded text-first interpretations:")
print(json.dumps(notes, indent=2, ensure_ascii=False))
```

## Reproduction and limitations

The source modules and staged commands are documented in the study README.
`exploration_codexgen` computes cached extents and closure;
`readingreplay_codexgen` runs bounded GPT-2 replay;
`prefixcontrast_codexgen` runs the input contrasts;
`structureprobe_codexgen` replays the fixed original MPS batches;
`semantictables_codexgen` regenerates the unified TeX tables.

No latent ablation, reconstruction-controlled behavioral test, independent
contextual annotation, or new corpus was added. Category mismatches do not
rule out contextual structure; exact numerical membership does not prove
that every returned item has a coherent semantic interpretation.
