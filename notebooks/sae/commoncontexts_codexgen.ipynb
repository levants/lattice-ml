block-15 MLP-output TopK SAE. Blocks and token positions are zero-based.

The full intent is the coordinatewise meet of five tokenwise joins.
Retrieval uses that entire vector at alpha=1. Both extents contain exactly
the five sources, and both intents pass the closure check. These are
separate canonical concepts, not a demonstrated cross-model infomorphism.

All five sources per condition are shown below; the paper shows the first
three. Yellow D means distributed satisfaction, not semantic correctness.
Rose marks any coordinate witness; bold would mark a whole-query witness.
