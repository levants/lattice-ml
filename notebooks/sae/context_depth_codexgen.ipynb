## Complete query inspection and the yellow convention

Retrieval uses the pooled item vector. A single token need not satisfy
all coordinates. The shared companion now checks every valid token
against the same scaled query used for item membership. S means at least
one whole-query token; yellow D means distributed satisfaction; R means
rejection. Rose marks any coordinate witness and bold marks whole-query
tokens. Yellow is not a semantic correctness annotation.

The appendix shows the entire cached span. The main paper uses one fixed
64-token window rule. The expandable ledger reports every coordinate's
requirement and maximal measured token activation. No score or selection
protocol is changed by this presentation update.
