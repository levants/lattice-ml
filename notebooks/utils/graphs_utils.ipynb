Comparing 2 vectors

````python
import matplotlib.pyplot as plt
import numpy as np

# Define two vectors
v1 = np.array([2, 5])
v2 = np.array([5, 8])

# Check if v1 <= v2 pointwise
is_less_equal = np.all(v1 <= v2)

# Create the plot
fig, ax = plt.subplots(figsize=(8, 8))

# Plot the vectors
ax.arrow(0, 0, v1[0], v1[1], head_width=0.3, head_length=0.5, fc='blue', ec='blue', length_includes_head=True, label=f'v1 = {v1}')
ax.arrow(0, 0, v2[0], v2[1], head_width=0.3, head_length=0.5, fc='red', ec='red', length_includes_head=True, label=f'v2 = {v2}')

# Add text labels for the vectors
ax.text(v1[0] * 0.7, v1[1] * 0.7, 'v1', color='blue', fontsize=14)
ax.text(v2[0] * 0.7, v2[1] * 0.7, 'v2', color='red', fontsize=14)

# Draw dashed lines from the vector tips to the axes for better visualization
ax.plot([v1[0], v1[0]], [0, v1[1]], 'b--', alpha=0.7)
ax.plot([0, v1[0]], [v1[1], v1[1]], 'b--', alpha=0.7)

ax.plot([v2[0], v2[0]], [0, v2[1]], 'r--', alpha=0.7)
ax.plot([0, v2[0]], [v2[1], v2[1]], 'r--', alpha=0.7)

# Highlight the pointwise comparison
if is_less_equal:
    title_text = "Pointwise Order: $v1 \leq v2$"
    comparison_text = f"Since {v1[0]} $\leq$ {v2[0]} and {v1[1]} $\leq$ {v2[1]}, then $v1 \leq v2$."
else:
    title_text = "Pointwise Order: $v1 \not\leq v2$"
    comparison_text = f"Pointwise comparison: ({v1[0]} $\leq$ {v2[0]}) and ({v1[1]} $\leq$ {v2[1]}) is False."

ax.set_title(title_text, fontsize=16)
ax.text(0.5, -0.1, comparison_text, transform=ax.transAxes, ha='center', fontsize=12, color='darkgreen')

# Set plot limits
max_coord = max(np.max(v1), np.max(v2)) + 1
ax.set_xlim(0, max_coord)
ax.set_ylim(0, max_coord)

# Add grid and labels
ax.grid(True, linestyle=':', alpha=0.7)
ax.set_xlabel('X-axis', fontsize=12)
ax.set_ylabel('Y-axis', fontsize=12)
ax.set_aspect('equal', adjustable='box')
ax.legend(fontsize=12)

plt.savefig(IMAGES / 'pointwise_vector_comparison.png', bbox_inches='tight')
plt.close(fig)
print("Image 'pointwise_vector_comparison.png' generated.")
````

Compare vectors

````python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Define two 3D vectors
v1_3d = np.array([2, 3, 4])
v2_3d = np.array([5, 6, 7])

# Check if v1 <= v2 pointwise
is_less_equal_3d = np.all(v1_3d <= v2_3d)

# Create the 3D plot
fig_3d = plt.figure(figsize=(10, 10))
ax_3d = fig_3d.add_subplot(111, projection='3d')

# Plot the vectors
# v1
ax_3d.quiver(0, 0, 0, v1_3d[0], v1_3d[1], v1_3d[2], color='blue', arrow_length_ratio=0.1, label=f'v1 = {v1_3d}')
# v2
ax_3d.quiver(0, 0, 0, v2_3d[0], v2_3d[1], v2_3d[2], color='red', arrow_length_ratio=0.1, label=f'v2 = {v2_3d}')

# Add text labels for the vectors
ax_3d.text(v1_3d[0] * 0.7, v1_3d[1] * 0.7, v1_3d[2] * 0.7, 'v1', color='blue', fontsize=14)
ax_3d.text(v2_3d[0] * 0.7, v2_3d[1] * 0.7, v2_3d[2] * 0.7, 'v2', color='red', fontsize=14)

# Draw dashed lines from the vector tips to the axes for better visualization
# For v1
ax_3d.plot([v1_3d[0], v1_3d[0]], [v1_3d[1], v1_3d[1]], [0, v1_3d[2]], 'b--', alpha=0.7) # Z projection
ax_3d.plot([v1_3d[0], v1_3d[0]], [0, v1_3d[1]], [v1_3d[2], v1_3d[2]], 'b--', alpha=0.7) # Y projection
ax_3d.plot([0, v1_3d[0]], [v1_3d[1], v1_3d[1]], [v1_3d[2], v1_3d[2]], 'b--', alpha=0.7) # X projection
ax_3d.plot([v1_3d[0], v1_3d[0]], [0, v1_3d[1]], [0, 0], 'b:', alpha=0.5) # Projection on XY plane
ax_3d.plot([0, v1_3d[0]], [v1_3d[1], v1_3d[1]], [0, 0], 'b:', alpha=0.5) # Projection on XY plane

# For v2
ax_3d.plot([v2_3d[0], v2_3d[0]], [v2_3d[1], v2_3d[1]], [0, v2_3d[2]], 'r--', alpha=0.7) # Z projection
ax_3d.plot([v2_3d[0], v2_3d[0]], [0, v2_3d[1]], [v2_3d[2], v2_3d[2]], 'r--', alpha=0.7) # Y projection
ax_3d.plot([0, v2_3d[0]], [v2_3d[1], v2_3d[1]], [v2_3d[2], v2_3d[2]], 'r--', alpha=0.7) # X projection
ax_3d.plot([v2_3d[0], v2_3d[0]], [0, v2_3d[1]], [0, 0], 'r:', alpha=0.5) # Projection on XY plane
ax_3d.plot([0, v2_3d[0]], [v2_3d[1], v2_3d[1]], [0, 0], 'r:', alpha=0.5) # Projection on XY plane


# Highlight the pointwise comparison
if is_less_equal_3d:
    title_text = "Pointwise Order in 3D: $v1 \leq v2$"
    comparison_text = f"Since {v1_3d[0]} $\leq$ {v2_3d[0]}, {v1_3d[1]} $\leq$ {v2_3d[1]}, and {v1_3d[2]} $\leq$ {v2_3d[2]}, then $v1 \leq v2$."
else:
    title_text = "Pointwise Order in 3D: $v1 \not\leq v2$"
    comparison_text = f"Pointwise comparison: ({v1_3d[0]} $\leq$ {v2_3d[0]}), ({v1_3d[1]} $\leq$ {v2_3d[1]}), and ({v1_3d[2]} $\leq$ {v2_3d[2]}) is False."

ax_3d.set_title(title_text, fontsize=16)
ax_3d.text2D(0.5, 0.05, comparison_text, transform=ax_3d.transAxes, ha='center', fontsize=12, color='darkgreen')


# Set plot limits
max_coord_3d = max(np.max(v1_3d), np.max(v2_3d)) + 1
ax_3d.set_xlim(0, max_coord_3d)
ax_3d.set_ylim(0, max_coord_3d)
ax_3d.set_zlim(0, max_coord_3d)

# Add grid and labels
ax_3d.set_xlabel('X-axis', fontsize=12)
ax_3d.set_ylabel('Y-axis', fontsize=12)
ax_3d.set_zlabel('Z-axis', fontsize=12)
ax_3d.grid(True, linestyle=':', alpha=0.7)
ax_3d.legend(fontsize=12)

# Adjust view angle for better visualization
ax_3d.view_init(elev=20, azim=135)

plt.savefig(IMAGES / 'pointwise_vector_comparison_3d.png', bbox_inches='tight')
plt.close(fig_3d)
print("Image 'pointwise_vector_comparison_3d.png' generated.")
````

````python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# --- Create 2D Pointwise Order Image ---

try:
    # Define the reference point
    P_2d = np.array([3, 4])
    
    # Set plot limits
    lim_min = 0
    lim_max = 10
    
    # Create the figure and axes
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Define the region for x' >= x and y' >= y
    x_region = [P_2d[0], lim_max, lim_max, P_2d[0]]
    y_region = [P_2d[1], P_2d[1], lim_max, lim_max]
    
    # Shade the region
    ax.fill(x_region, y_region, color='lightblue', alpha=0.6, label=r'Region $v \geq P$')
    
    # Plot the reference point
    ax.plot(P_2d[0], P_2d[1], 'ro', markersize=10, label=r'$P = (3, 4)$')
    
    # Plot an arrow from origin to P
    ax.arrow(0, 0, P_2d[0], P_2d[1], head_width=0.2, head_length=0.3, fc='gray', ec='gray', length_includes_head=True)
    ax.text(P_2d[0] * 0.5, P_2d[1] * 0.5, 'P', color='black', fontsize=12)

    # Draw dashed lines from P to the axes
    ax.plot([P_2d[0], P_2d[0]], [lim_min, P_2d[1]], 'r--')
    ax.plot([lim_min, P_2d[0]], [P_2d[1], P_2d[1]], 'r--')
    
    # Set labels, title, and limits
    ax.set_xlabel('$x$-axis', fontsize=14)
    ax.set_ylabel('$y$-axis', fontsize=14)
    ax.set_title('Pointwise Order in 2D', fontsize=16)
    ax.set_xlim(lim_min, lim_max)
    ax.set_ylim(lim_min, lim_max)
    
    # Add grid
    ax.grid(True, linestyle=':', alpha=0.7)
    
    # Add legend
    ax.legend(fontsize=12)
    
    # Add text
    ax.text(6.5, 7, r"Shaded region represents all vectors $v = (x', y')$" "\n" r"such that $x' \geq 3$ and $y' \geq 4$",
            fontsize=12, ha='center', bbox=dict(facecolor='white', alpha=0.5, boxstyle='round,pad=0.5'))
    
    # Ensure equal aspect ratio
    ax.set_aspect('equal', adjustable='box')
    
    # Save the figure
    plt.savefig(IMAGES / 'pointwise_order_2d.png', bbox_inches='tight')
    plt.close(fig) # Close the figure to free up memory
    print("2D pointwise order image saved as 'pointwise_order_2d.png'")

except Exception as e:
    print(f"Error creating 2D plot: {e}")

# --- Create 3D Pointwise Order Image ---

try:
    # Define the reference point
    P_3d = np.array([2, 3, 4])
    
    # Set plot limits
    lim_min_3d = 0
    lim_max_3d = 8
    
    # Create the figure and 3D axes
    fig_3d = plt.figure(figsize=(10, 10))
    ax_3d = fig_3d.add_subplot(111, projection='3d')
    
    # Plot the reference point
    ax_3d.scatter(P_3d[0], P_3d[1], P_3d[2], c='r', marker='o', s=100, label=r'$P = (2, 3, 4)$')
    
    # Define the surfaces representing the boundaries of the region
    # We create a mesh for the planes starting from P up to the plot limit
    
    # Plane 1: x = P_x (y-z plane at x=2) for y >= 3, z >= 4
    yy1, zz1 = np.meshgrid(np.linspace(P_3d[1], lim_max_3d, 10), 
                           np.linspace(P_3d[2], lim_max_3d, 10))
    xx1 = np.full_like(yy1, P_3d[0])
    
    # Plane 2: y = P_y (x-z plane at y=3) for x >= 2, z >= 4
    xx2, zz2 = np.meshgrid(np.linspace(P_3d[0], lim_max_3d, 10), 
                           np.linspace(P_3d[2], lim_max_3d, 10))
    yy2 = np.full_like(xx2, P_3d[1])
    
    # Plane 3: z = P_z (x-y plane at z=4) for x >= 2, y >= 3
    xx3, yy3 = np.meshgrid(np.linspace(P_3d[0], lim_max_3d, 10), 
                           np.linspace(P_3d[1], lim_max_3d, 10))
    zz3 = np.full_like(xx3, P_3d[2])
    
    # Plot the surfaces
    ax_3d.plot_surface(xx1, yy1, zz1, alpha=0.3, color='blue')
    ax_3d.plot_surface(xx2, yy2, zz2, alpha=0.3, color='green')
    ax_3d.plot_surface(xx3, yy3, zz3, alpha=0.3, color='orange')
    
    # Draw dashed lines from P to the base planes
    ax_3d.plot([P_3d[0], P_3d[0]], [P_3d[1], P_3d[1]], [lim_min_3d, P_3d[2]], 'r--')
    ax_3d.plot([P_3d[0], P_3d[0]], [lim_min_3d, P_3d[1]], [P_3d[2], P_3d[2]], 'r--')
    ax_3d.plot([lim_min_3d, P_3d[0]], [P_3d[1], P_3d[1]], [P_3d[2], P_3d[2]], 'r--')
    
    # Draw projections on the base planes
    ax_3d.scatter(P_3d[0], P_3d[1], lim_min_3d, c='gray', marker='x') # projection on xy
    ax_3d.scatter(P_3d[0], lim_min_3d, P_3d[2], c='gray', marker='x') # projection on xz
    ax_3d.scatter(lim_min_3d, P_3d[1], P_3d[2], c='gray', marker='x') # projection on yz

    # Add text
    ax_3d.text(5, 5.5, 6, r"The surfaces define the boundaries" "\n" r"of the region $v \geq P$, where" "\n" r"$v = (x', y', z')$ has $x' \geq 2$, $y' \geq 3$, $z' \geq 4$.",
               fontsize=10, ha='center', bbox=dict(facecolor='white', alpha=0.5, boxstyle='round,pad=0.5'))

    # Set labels, title, and limits
    ax_3d.set_xlabel('$x$-axis', fontsize=12)
    ax_3d.set_ylabel('$y$-axis', fontsize=12)
    ax_3d.set_zlabel('$z$-axis', fontsize=12)
    ax_3d.set_title('Pointwise Order in 3D', fontsize=16)
    ax_3d.set_xlim(lim_min_3d, lim_max_3d)
    ax_3d.set_ylim(lim_min_3d, lim_max_3d)
    ax_3d.set_zlim(lim_min_3d, lim_max_3d)
    
    # Add legend
    ax_3d.legend(fontsize=12)
    
    # Adjust view angle
    ax_3d.view_init(elev=20, azim=135)
    
    # Save the figure
    plt.savefig(IMAGES / 'pointwise_order_3d.png', bbox_inches='tight')
    plt.close(fig_3d) # Close the figure
    print("3D pointwise order image saved as 'pointwise_order_3d.png'")

except Exception as e:
    print(f"Error creating 3D plot: {e}")
````
